# Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.plot([1, 2, 3, 4], [2, 3, 6, 7])
```

```python
n = np.arange(0, 5, 0.3)
plt.plot(n, n, "ro", n, n**2, "gs", n, n**3, "b--")
```

```python
data = {
    'a': np.arange(500),
    'b': np.random.randint(0, 50, 50),
    'c': np.random.randn(500)
}
plt.scatter(data['a'], data['c'])
```
